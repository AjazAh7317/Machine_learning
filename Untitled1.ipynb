# VALUE IMPUTATION

```python
from sklearn.impute import SimpleImputer
```

```python
print(df.age.isnull().sum())
```

```python
imp = SimpleImputer(strategy='mean')
df['age'] = imp.fit_transform(df[['age']])
print(f'Number of null values in age column after the imputation: {df.age.isnull().sum()}')
```

```python
def get_parameter(df):
    Parameters = {}

    for col in df.columns[df.isnull().any()]:
        if df[col].dtype == 'float64' or df[col].dtype == 'int64' or df[col].dtype == 'int32':
            strategy = 'mean'
        else:
            strategy = 'most_frequent'

        Parameters[col] = strategy

    return Parameters


get_parameter(df)
```

```python
parameters=get_parameter(df)
```

```python
parameters
```

```python
for col, strategy in parameters.items():
    imp = SimpleImputer(strategy=strategy)
    # .ravel() flattens the 2D output of fit_transform back to a 1D array
    df[col] = imp.fit_transform(df[[col]]).ravel()
```
